```python
import yfinance as yf
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Define the assets you want to track
# Brent Oil, US 10Y Yield, Nifty 50 (India), and Apple (US Tech)
tickers = ["BZ=F", "^TNX", "^NSEI", "AAPL", "GOOGL"]

# 2. Download the data (last 2 years)
df = yf.download(tickers, period="2y")['Close']

# 3. Convert prices to daily percentage changes (Returns)
# This is crucial! Correlations on raw prices are often misleading.
returns = df.pct_change().dropna()

# 4. Create the correlation matrix
matrix = returns.corr()

# 5. Visualize it with a Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(matrix, annot=True, cmap='RdYlGn', center=0)
plt.title('Asset Correlation Matrix: Macro vs. Equity')
plt.show()
```
